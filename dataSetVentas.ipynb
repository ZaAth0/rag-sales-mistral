Cargar el DataSet

```python
import pandas as pd
import numpy as np
from google.colab import files
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import chardet

# Paso 1: Detectar y cargar archivo
def cargar_csv_seguro(nombre_archivo):
    # Probar codificaciones comunes
    codificaciones = ['ISO-8859-1', 'latin-1', 'cp1252', 'utf-8', 'utf-16']

    for encoding in codificaciones:
        try:
            df_temp = pd.read_csv(nombre_archivo, encoding=encoding)
            print(f"✓ Cargado exitosamente con {encoding}")
            return df_temp, encoding
        except:
            continue

    # Si ninguna funciona, usar detección automática
    with open(nombre_archivo, 'rb') as f:
        result = chardet.detect(f.read())
        encoding_auto = result['encoding']
        df_temp = pd.read_csv(nombre_archivo, encoding=encoding_auto)
        print(f"✓ Cargado con detección automática: {encoding_auto}")
        return df_temp, encoding_auto

# Cargar tu archivo
df, encoding_usado = cargar_csv_seguro('sales_data_sample.csv')

# Verificar resultado
print(f"\nDimensiones del dataset: {df.shape}")
print(f"\nColumnas disponibles:")
print(df.columns.tolist())
print(f"\nPrimeras 3 filas:")
print(df.head(3))
```

Validacion y Limpieza

```python
# Paso 2: Exploración inicial
print("="*50)
print("INFORMACIÓN DEL DATASET")
print("="*50)
print(f"Total registros: {len(df)}")
print(f"Total columnas: {len(df.columns)}")
print(f"\nTipos de datos:")
print(df.dtypes.value_counts())

# Paso 3: Verificar valores nulos
print(f"\nVALORES NULOS:")
nulos = df.isnull().sum()
if nulos.sum() > 0:
    print(nulos[nulos > 0])
else:
    print("✓ No hay valores nulos")

# Paso 4: Limpiar NaN si existen
df_clean = df.dropna()
print(f"\nRegistros después de limpiar NaN: {len(df_clean)}")

# Paso 5: Identificar columnas numéricas
numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
print(f"\nCOLUMNAS NUMÉRICAS ({len(numeric_cols)}):")
print(numeric_cols.tolist())

# Paso 6: Ver estadísticas básicas
print(f"\nESTADÍSTICAS DE DATOS NUMÉRICOS:")
print(df_clean[numeric_cols].describe())

# Paso 7: Normalizar (solo columnas numéricas)
if len(numeric_cols) > 0:
    scaler = MinMaxScaler()
    df_normalized = df_clean.copy()
    df_normalized[numeric_cols] = scaler.fit_transform(df_clean[numeric_cols])

    # Verificar normalización
    print(f"\nNORMALIZACIÓN COMPLETADA:")
    for col in numeric_cols[:5]:  # Mostrar primeras 5
        print(f"{col}: min={df_normalized[col].min():.3f}, max={df_normalized[col].max():.3f}")

    # Guardar resultado
    df_normalized.to_csv('sales_data_normalizado.csv', index=False)
    print(f"\n✓ Dataset normalizado guardado como 'sales_data_normalizado.csv'")
    print(f"  Forma final: {df_normalized.shape}")

    # Descargar
    files.download('sales_data_normalizado.csv')
else:
    print("⚠ No hay columnas numéricas para normalizar")
```
